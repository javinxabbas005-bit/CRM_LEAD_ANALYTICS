```python
import pandas as p
```

```python
d=p.read_csv("olist_closed_deals_dataset.csv")
```

```python
l=p.read_csv("olist_marketing_qualified_leads_dataset.csv")
```

```python
d.head()
```

```python
l.head()
```

```python
combine=l.merge(d,on="mql_id",how="inner")
originleads=l["origin"].value_counts()
origindeals=combine["origin"].value_counts()
rateofconv=(origindeals/originleads*100).round(2)
rateofconv.sort_values(ascending=False)
```

```python
import matplotlib.pyplot as pl
```

```python
rateofconv.sort_values(ascending=False).plot(kind="bar",figsize=(11,5),color="lavender")
pl.title("Conversion Rate by Marketing Channel")
pl.ylabel("Conversion Rate (%)")
pl.xlabel("origin")
pl.tight_layout()
pl.show()                                          
```
